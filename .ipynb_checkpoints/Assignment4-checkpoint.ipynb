## Setup and Environment

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

PRICES = [5, 10, 15, 20, 25]
TRUE_CONVERSION_RATES = [0.50, 0.35, 0.22, 0.12, 0.05]

def simulate_sale(price_index):
    """Simulate one customer. Returns revenue (price if they buy, 0 if not)."""
    bought = np.random.random() < TRUE_CONVERSION_RATES[price_index]
    return PRICES[price_index] if bought else 0

print("Environment loaded")
print(f"Prices: {PRICES}")
```

## Part 1: Multi-Armed Bandits

Three strategies for finding the best price:
- **Random**: No learning
- **ε-Greedy**: Usually exploit the best, sometimes explore
- **UCB**: Balance exploration and exploitation using confidence bounds

```python
def run_bandit(strategy, n_customers, epsilon=0.1, ucb_confidence=2.0):
    """Run a bandit algorithm."""
    counts = [0] * 5
    total_rewards = [0] * 5
    cumulative = 0
    history = []
    
    for t in range(n_customers):
        if strategy == "random":
            choice = np.random.randint(5)
        elif strategy == "epsilon-greedy":
            if min(counts) == 0:
                choice = counts.index(0)
            elif np.random.random() < epsilon:
                choice = np.random.randint(5)
            else:
                avgs = [total_rewards[i] / counts[i] for i in range(5)]
                choice = avgs.index(max(avgs))
        elif strategy == "ucb":
            if min(counts) == 0:
                choice = counts.index(0)
            else:
                ucb_scores = []
                for i in range(5):
                    avg = total_rewards[i] / counts[i]
                    bonus = ucb_confidence * np.sqrt(np.log(t + 1) / counts[i])
                    ucb_scores.append(avg + bonus)
                choice = ucb_scores.index(max(ucb_scores))
        
        revenue = simulate_sale(choice)
        counts[choice] += 1
        total_rewards[choice] += revenue
        cumulative += revenue
        history.append(cumulative)
    
    return history, counts

print("Bandit functions loaded")
```

## Experiment: Bandit Strategies

```python
# ============================================================
# EXPERIMENT: Try different epsilon values
# ============================================================
EPSILON = 0.1  # Try: 0.0, 0.1, 0.3, 0.5
N_CUSTOMERS = 1000

random_history, random_counts = run_bandit("random", N_CUSTOMERS)
eg_history, eg_counts = run_bandit("epsilon-greedy", N_CUSTOMERS, epsilon=EPSILON)
ucb_history, ucb_counts = run_bandit("ucb", N_CUSTOMERS)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(random_history, label="Random", alpha=0.7)
axes[0].plot(eg_history, label=f"ε-Greedy (ε={EPSILON})", linewidth=2)
axes[0].plot(ucb_history, label="UCB", linewidth=2)
axes[0].set_xlabel("Customer #")
axes[0].set_ylabel("Cumulative Revenue ($)")
axes[0].set_title("Revenue Over Time")
axes[0].legend()

x = np.arange(5)
width = 0.25
axes[1].bar(x - width, random_counts, width, label="Random", alpha=0.7)
axes[1].bar(x, eg_counts, width, label="ε-Greedy", alpha=0.7)
axes[1].bar(x + width, ucb_counts, width, label="UCB", alpha=0.7)
axes[1].set_xticks(x)
axes[1].set_xticklabels(["$5", "$10", "$15", "$20", "$25"])
axes[1].set_xlabel("Price")
axes[1].set_ylabel("Times Chosen")
axes[1].set_title("Price Selection Frequency")
axes[1].legend()

plt.tight_layout()
plt.show()

print(f"Final Revenue after {N_CUSTOMERS} customers:")
print(f"  Random:    ${random_history[-1]:,}")
print(f"  ε-Greedy:  ${eg_history[-1]:,}")
print(f"  UCB:       ${ucb_history[-1]:,}")
```

## Part 2: Q-Learning with State

Q-learning adds **state awareness**. The optimal price now depends on:
- **Inventory level**: High vs Low
- **Time remaining**: Early vs Late in season

Key insight: Low inventory = scarcity = can charge premium!

```python
class PricingGame:
    """Seasonal pricing game where optimal price depends on state."""
    
    def __init__(self, inventory=15, periods=20):
        self.start_inv = inventory
        self.total_periods = periods
        self.reset()
    
    def reset(self):
        self.inventory = self.start_inv
        self.periods_left = self.total_periods
        return self.get_state()
    
    def get_state(self):
        inv = "high" if self.inventory > 8 else "low"
        time = "early" if self.periods_left > 10 else "late"
        return (inv, time)
    
    def step(self, price_index):
        if self.periods_left <= 0 or self.inventory <= 0:
            return self.get_state(), 0, True
        
        inv, time = self.get_state()
        
        # State-dependent conversion rates
        rates = {
            ("high", "early"): [0.85, 0.65, 0.35, 0.12, 0.04],
            ("high", "late"):  [0.92, 0.80, 0.55, 0.22, 0.07],
            ("low", "early"):  [0.45, 0.50, 0.48, 0.38, 0.22],
            ("low", "late"):   [0.65, 0.70, 0.68, 0.52, 0.32],
        }
        
        rate = rates[(inv, time)][price_index]
        sold = np.random.random() < rate
        revenue = PRICES[price_index] if sold else 0
        
        if sold:
            self.inventory -= 1
        self.periods_left -= 1
        
        done = self.periods_left <= 0 or self.inventory <= 0
        
        if done and self.inventory > 0:
            revenue -= self.inventory * 4
        
        return self.get_state(), revenue, done

def run_qlearning(n_episodes, learning_rate=0.15, discount=0.9):
    """Train a Q-learning agent."""
    states = [("high", "early"), ("high", "late"), ("low", "early"), ("low", "late")]
    Q = {s: [0.0] * 5 for s in states}
    
    episode_revenues = []
    
    for episode in range(n_episodes):
        epsilon = max(0.05, 1.0 - episode / (n_episodes * 0.7))
        
        game = PricingGame()
        state = game.reset()
        total_revenue = 0
        
        while True:
            if np.random.random() < epsilon:
                action = np.random.randint(5)
            else:
                action = Q[state].index(max(Q[state]))
            
            next_state, revenue, done = game.step(action)
            total_revenue += revenue
            
            old_value = Q[state][action]
            target = revenue if done else revenue + discount * max(Q[next_state])
            Q[state][action] = old_value + learning_rate * (target - old_value)
            
            state = next_state
            if done:
                break
        
        episode_revenues.append(total_revenue)
    
    return episode_revenues, Q

print("Q-Learning functions loaded")
```

## Experiment: Q-Learning

```python
# ============================================================
# EXPERIMENT: Try different parameters
# ============================================================
LEARNING_RATE = 0.15  # Try: 0.01, 0.1, 0.3, 0.5
DISCOUNT = 0.9        # Try: 0.0, 0.5, 0.9, 0.99
N_EPISODES = 1500

revenues, Q_table = run_qlearning(N_EPISODES, LEARNING_RATE, DISCOUNT)

fig, ax = plt.subplots(figsize=(10, 5))

window = 50
smoothed = [np.mean(revenues[max(0, i-window):i+1]) for i in range(len(revenues))]

ax.plot(revenues, alpha=0.2, color="blue")
ax.plot(smoothed, color="blue", linewidth=2, label="Average (smoothed)")
ax.axhline(y=np.mean(revenues[:50]), color="red", linestyle="--", alpha=0.7, 
           label=f"Early avg: ${np.mean(revenues[:50]):.0f}")
ax.axhline(y=np.mean(revenues[-50:]), color="green", linestyle="--", alpha=0.7, 
           label=f"Final avg: ${np.mean(revenues[-50:]):.0f}")
ax.set_xlabel("Episode")
ax.set_ylabel("Total Revenue ($)")
ax.set_title(f"Q-Learning (LR={LEARNING_RATE}, Discount={DISCOUNT})")
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()

improvement = np.mean(revenues[-50:]) - np.mean(revenues[:50])
pct = improvement / np.mean(revenues[:50]) * 100
print(f"Average revenue (first 50 episodes):  ${np.mean(revenues[:50]):.1f}")
print(f"Average revenue (last 50 episodes):   ${np.mean(revenues[-50:]):.1f}")
print(f"Improvement: ${improvement:.1f} ({pct:.0f}% increase)")
```

## Learned Policy

```python
fig, ax = plt.subplots(figsize=(8, 5))

policy_grid = []
for inv in ["high", "low"]:
    row = []
    for time in ["early", "late"]:
        best_action = Q_table[(inv, time)].index(max(Q_table[(inv, time)]))
        row.append(PRICES[best_action])
    policy_grid.append(row)

im = ax.imshow(policy_grid, cmap="RdYlGn_r", vmin=5, vmax=25)

ax.set_xticks([0, 1])
ax.set_xticklabels(["Early Season", "Late Season"])
ax.set_yticks([0, 1])
ax.set_yticklabels(["High Inventory", "Low Inventory"])

for i in range(2):
    for j in range(2):
        ax.text(j, i, f"${policy_grid[i][j]}", ha="center", va="center", 
                fontsize=24, fontweight="bold")

plt.title("Learned Policy: Best Price by Situation", fontsize=14)
plt.colorbar(im, label="Price ($)")
plt.tight_layout()
plt.show()
```

## Random vs Learned Policy

```python
def run_learned_policy(Q, n_episodes):
    revenues = []
    for _ in range(n_episodes):
        game = PricingGame()
        state = game.reset()
        total = 0
        while True:
            action = Q[state].index(max(Q[state]))
            state, revenue, done = game.step(action)
            total += revenue
            if done:
                break
        revenues.append(total)
    return revenues

def run_random_policy(n_episodes):
    revenues = []
    for _ in range(n_episodes):
        game = PricingGame()
        state = game.reset()
        total = 0
        while True:
            action = np.random.randint(5)
            state, revenue, done = game.step(action)
            total += revenue
            if done:
                break
        revenues.append(total)
    return revenues

learned_revenues = run_learned_policy(Q_table, 500)
random_revenues = run_random_policy(500)

fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(["Random Policy", "Learned Policy"], 
       [np.mean(random_revenues), np.mean(learned_revenues)],
       yerr=[np.std(random_revenues)/5, np.std(learned_revenues)/5],
       color=["#ff6b6b", "#45b7d1"], edgecolor="k", capsize=5)
ax.set_ylabel("Average Revenue ($)")
ax.set_title("Random vs Learned Policy")

plt.tight_layout()
plt.show()

improvement = np.mean(learned_revenues) - np.mean(random_revenues)
pct = improvement / np.mean(random_revenues) * 100
print(f"Random policy:  ${np.mean(random_revenues):.1f}")
print(f"Learned policy: ${np.mean(learned_revenues):.1f}")
print(f"Improvement: ${improvement:.1f} ({pct:.0f}% better)")
```

## Compare Learning Rates

```python
learning_rates = [0.01, 0.1, 0.3, 0.5]

fig, ax = plt.subplots(figsize=(10, 5))

for lr in learning_rates:
    revs, _ = run_qlearning(1500, learning_rate=lr)
    smoothed = [np.mean(revs[max(0, i-50):i+1]) for i in range(len(revs))]
    ax.plot(smoothed, label=f"LR = {lr}", linewidth=2)

ax.set_xlabel("Episode")
ax.set_ylabel("Revenue (smoothed)")
ax.set_title("Effect of Learning Rate")
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
