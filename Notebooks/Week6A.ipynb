# Week 6A: Word Embeddings (GloVe)

# 1 Setup

#### Import libraries for word embeddings and similarity analysis.

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

from sklearn.decomposition import PCA
from sklearn.metrics.pairwise import cosine_similarity

import gensim.downloader as api
import matplotlib.pyplot as plt
```

---
# 2 Load GloVe Embeddings

#### Load pre-trained GloVe embeddings using gensim. These 50-dimensional vectors are trained on Wikipedia and capture semantic relationships.

```python
print("Downloading GloVe embeddings (this may take a minute)...")
glove = api.load('glove-wiki-gigaword-50')
print(f"Vocabulary size: {len(glove):,} words")
print(f"Embedding dimension: {glove.vector_size}")
print(f"\nExample — vector for 'king' (first 10 dimensions):")
print(glove['king'][:10].round(4))
```

---
# 3 Word Similarity

#### Compute cosine similarity between word pairs and find nearest neighbors. Words used in similar contexts cluster together in embedding space.

```python
# Compute cosine similarity between word pairs
test_pairs = [
    ('king', 'queen'),
    ('king', 'banana'),
    ('happy', 'joyful'),
    ('happy', 'sad'),
    ('dog', 'cat'),
    ('dog', 'computer'),
    ('paris', 'france'),
    ('tokyo', 'japan'),
]

print("Word Pair Similarities:\n")
similarities = []
for w1, w2 in test_pairs:
    sim = glove.similarity(w1, w2)
    similarities.append(sim)
    print(f"  {w1:10s} ~ {w2:10s}: {sim:.4f}")
```

```python
# Plot word similarity scores
import matplotlib.pyplot as plt

pair_labels = [f"{w1}\n{w2}" for w1, w2 in test_pairs]
plt.figure(figsize=(12, 6))
plt.bar(range(len(similarities)), similarities, color='steelblue')
plt.xticks(range(len(test_pairs)), pair_labels, rotation=45, ha='right', fontsize=9)
plt.ylabel('Cosine Similarity')
plt.title('Word Pair Similarity Scores (GloVe Embeddings)')
plt.tight_layout()
plt.show()
```

```python
# Find the most similar words to a given word
query_words = ['king', 'computer', 'happy', 'ocean']

print("\nMost Similar Words:\n")
for word in query_words:
    similar = glove.most_similar(word, topn=5)
    neighbors = ', '.join([f"{w} ({s:.2f})" for w, s in similar])
    print(f"  {word:10s} → {neighbors}")
```

# 4 Word Analogies

#### Solve word analogies using vector arithmetic. The relationship 'king - man + woman' shows how word vectors encode semantic structure algebraically.

```python
# Solve word analogies: a is to b as c is to ?
analogies = [
    ('king',   'man',    'woman'),    # royalty gender
    ('paris',  'france', 'japan'),    # capital-country
    ('bigger', 'big',    'small'),    # comparative
    ('walking','walk',   'swim'),     # verb tense
]

print("Word Analogies: a - b + c = ?\n")
for a, b, c in analogies:
    results = glove.most_similar(positive=[a, c], negative=[b], topn=3)
    answers = ', '.join([f"{w} ({s:.2f})" for w, s in results])
    print(f"  {a} - {b} + {c}")
    print(f"    → {answers}\n")
```
