## How the Simulation Works

You are an airline selling seats on a series of flights. Each flight has:
- **75 seats** available
- **180-day** booking window before departure
- **1 customer per day** arrives and decides whether to buy

### Customer Behavior
Each customer has a **willingness-to-pay (WTP)** that increases over time:
- Early bookers (day 1-50): WTP around $100-150
- Mid-period (day 50-120): WTP around $150-250  
- Late bookers (day 150-180): WTP can exceed $500

If your price ≤ customer's WTP, they buy. Otherwise, they leave.

### Your Goal
Maximize total revenue across 20 flights by choosing the right pricing strategy.

---

## The Three Pricing Strategies

### 📊 Static Pricing
Set a single fixed price for all customers on all flights. Use the slider to choose your price.

### 🎰 Bandit (Thompson Sampling)
The algorithm learns which single price works best by experimenting:
- Each flight, it picks a price to try
- It observes the revenue and updates its beliefs
- Over time, it converges to the best static price

### 🧠 Q-Learning
The algorithm learns to price *dynamically* based on the current situation:
- **State** = (capacity remaining) × (time until departure)
- It can learn different optimal prices for different states
- For example: price low when seats are plentiful, price high near departure
