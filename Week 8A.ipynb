## Part 2: The Art of the Prompt: Generating a Product Description

The most important skill in using Generative AI is **prompt engineering**. The model is powerful, but it needs clear, detailed instructions. Let's see the difference between a weak prompt and a strong one.

```python
product_name = "The 'EverWarm' Smart Mug"
product_features = "12oz capacity, self-heating, 3-hour battery life, ceramic finish."

# --- Weak Prompt ---
weak_prompt = f"Write a product description for {product_name}."
print("--- Generating with a WEAK prompt ---")
response = model.generate_content(weak_prompt)
print(response.text)

# --- Strong Prompt ---
# A strong prompt includes: Role, Task, Context, Format, and Tone.
strong_prompt = f"""
**Role**: Act as an expert e-commerce copywriter.
**Task**: Write a compelling 100-word product description.
**Context**: The product is called '{product_name}'. Its key features are: {product_features}.
**Format**: A single paragraph.
**Tone**: Aspirational and tech-focused.
**Key Benefit**: The main benefit to highlight is the joy of never drinking cold coffee again.
"""
print("\n--- Generating with a STRONG prompt ---")
response = model.generate_content(strong_prompt)
print(response.text)
```

The difference is clear. The strong prompt gives us a much more polished, targeted, and useful result because we provided specific instructions and context.

## Part 3: Few-Shot Prompting: Creating Ad Copy

To get even more control, especially over the output format, we can provide the model with examples. This is called **few-shot** or **one-shot prompting**. We'll show it an example of good ad copy, then ask it to generate new copy in the same style.

```python
few_shot_prompt = f"""
I need to create ad copy in a specific format: a headline, a body paragraph, and a call to action.

-- EXAMPLE --
**Product**: 'Silent-Key' Mechanical Keyboard
**Ad Copy**:
**Headline**: Find Your Focus. Hear Only Your Thoughts.
**Body**: The Silent-Key Mechanical Keyboard combines the satisfying feel of a mechanical switch with an ultra-quiet design. Perfect for open offices and late-night work sessions.
**Call to Action**: Upgrade Your Workspace Today.
------------- 

-- NEW TASK --
**Product**: {product_name}
**Ad Copy**:
"""

print("--- Generating with a FEW-SHOT prompt ---")
response = model.generate_content(few_shot_prompt)
print(response.text)
```

By providing an example, we constrained the model to follow our desired structure perfectly. This is a powerful technique for generating consistent, reliable outputs.

---

## Part 4: Building Reusable Tools

Finally, let's make our assistant more useful by wrapping the logic in a Python function. This allows us to easily generate different kinds of content without rewriting the API call every time.

```python
def marketing_assistant(prompt):
  """A simple function to send a prompt to the Gemini model and return the text."""
  try:
    response = model.generate_content(prompt)
    return response.text
  except Exception as e:
    return f"An error occurred: {e}"

# Now let's use our function for a new task
instagram_prompt = f"""
Act as a social media manager. Write 3 distinct, engaging Instagram captions for the '{product_name}'. 
Each caption should be under 50 words and include relevant emojis and hashtags.
"""

print("--- Generating Instagram Captions ---")
instagram_captions = marketing_assistant(instagram_prompt)
print(instagram_captions)

email_prompt = f"""
Act as an email marketer. Write 5 compelling subject lines for a launch email about the '{product_name}'.
The subject lines should create curiosity and highlight the main benefit.
"""

print("\n--- Generating Email Subject Lines ---")
email_subjects = marketing_assistant(email_prompt)
print(email_subjects)
```
